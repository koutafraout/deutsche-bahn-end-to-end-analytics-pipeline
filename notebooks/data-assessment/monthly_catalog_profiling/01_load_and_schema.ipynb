## Convert timestamp
A timestamp such as: 2024-07-01 12:30:00 is stored in the Parquet file as nanoseconds: 1719837000000000000

Spark provides timestamp_micros() for converting epoch microseconds to a Spark timestamp. Because the source values are nanoseconds, they must first be divided by 1,000.


```python
# Use UTC internally for consistent pipeline processing.
spark.conf.set("spark.sql.session.timeZone", "UTC")

timestamp_columns = [
    "time",
    "arrival_planned_time",
    "arrival_change_time",
    "departure_planned_time",
    "departure_change_time",
]

historical_df_converted = historical_df

for column_name in timestamp_columns:
    historical_df_converted = historical_df_converted.withColumn(
        f"{column_name}_converted",
        F.when(
            F.col(column_name).isNotNull(),
            F.timestamp_micros(
                (F.col(column_name) / F.lit(1_000)).cast("long")
            ),
        ).otherwise(F.lit(None).cast("timestamp")),
    )
```

```python
# Inspect the converted values:
historical_df_converted.select(
    "station_name",
    "xml_station_name",
    "time",
    "time_converted",
    "arrival_planned_time",
    "arrival_planned_time_converted",
    "arrival_change_time",
    "arrival_change_time_converted",
    "departure_planned_time",
    "departure_planned_time_converted",
    "departure_change_time",
    "departure_change_time_converted",
).show(
    1,
    truncate=False,
    vertical=True,
)
```

The source timestamps represent Europe/Berlin local time and contain no
timezone information.

Spark reads the Parquet nanosecond timestamps as long integers through the
compatibility option. The Spark session is temporarily fixed to UTC so that
conversion with `timestamp_micros()` is deterministic and preserves the
source wall-clock values without applying an additional machine-dependent
timezone offset.

The resulting values still represent Europe/Berlin local time; they should
not be interpreted as timestamps originally recorded in UTC.

```python
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found at {DATA_PATH}. "
        "Download the monthly Parquet sample first."
    )

historical_df_raw = spark.read.parquet(str(DATA_PATH))

timestamp_columns = [
    "time",
    "arrival_planned_time",
    "arrival_change_time",
    "departure_planned_time",
    "departure_change_time",
]

historical_df = historical_df_raw

for column_name in timestamp_columns:
    if column_name in historical_df.columns:
        historical_df = historical_df.withColumn(
            column_name,
            F.when(
                F.col(column_name).isNotNull(),
                F.timestamp_micros(
                    (
                        F.col(column_name)
                        / F.lit(1_000)
                    ).cast("long")
                ),
            ).otherwise(
                F.lit(None).cast("timestamp")
            ),
        )

historical_df.cache()

print("Raw schema:")
historical_df_raw.printSchema()

print("\nConverted schema:")
historical_df.printSchema()
```
